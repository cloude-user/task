## 1. Variables, Scope & Data Types
| Python | JavaScript | Rule |
| :--- | :--- | :--- |
| `name = "value"` | `const name = "value";` | Use `const` by default (immutable binding) |
| `count = 0; count += 1` | `let count = 0; count += 1;` | Use `let` for variables that change |
| `None` | `null` / `undefined` | `null` is explicit empty; `undefined` is unassigned |
| `True` / `False` | `true` / `false` | Lowercase booleans in JS |


````python
# Python Demo Code
server_name = "nalgonda-web"
is_running = True
server_count = 3
server_count += 1
print(f"Python: {server_name}, Running: {is_running}, Count: {server_count}")
````

## 2. Arrow Functions vs Python Lambdas
In Python, lambdas are restricted to 1 line. In JavaScript, **Arrow Functions (`=>`)** are used everywhere—both for 1-line expressions and full blocks!

| Feature | Python | JavaScript |
| :--- | :--- | :--- |
| **Normal Function** | `def add(a, b): return a + b` | `function add(a, b) { return a + b; }` |
| **Inline Lambda / Arrow** | `lambda a, b: a + b` | `(a, b) => a + b` |
| **Multi-line Arrow** | *(Not supported in Python lambda)* | `(a, b) => { const sum = a + b; return sum; }` |


````python
# Python: Filtering and mapping
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Python doubled:", doubled)
print("Python evens:", evens)
````

### The JavaScript Equivalent:
```javascript
const numbers = [1, 2, 3, 4, 5];

// Map with arrow function
const doubled = numbers.map(x => x * 2);

// Filter with arrow function
const evens = numbers.filter(x => x % 2 === 0);

console.log("JS doubled:", doubled);
console.log("JS evens:", evens);
```

## 3. Asynchronous Programming: Promises and `async / await`
In Cloud APIs (like AWS S3 or DynamoDB), network calls take time.
* Python uses sequential blocking code by default (or `asyncio`).
* JavaScript uses an **asynchronous event loop** with **Promises** and **`async / await`** so the server never freezes while waiting for a response!


### Visualizing a Cloud API Call in JavaScript:
```javascript
// Simulating an AWS S3 Fetch in Node.js
function getS3Object(bucket, key) {
    return new Promise((resolve, reject) => {
        setTimeout(() => {
            if (bucket && key) {
                resolve({ statusCode: 200, data: `Contents of s3://${bucket}/${key}` });
            } else {
                reject(new Error("Missing bucket or key!"));
            }
        }, 1000); // 1-second simulated AWS network latency
    });
}

// Consuming with Modern async / await (Standard in AWS Lambda Node.js)
async function handler() {
    try {
        console.log("⏳ Requesting data from AWS S3...");
        const response = await getS3Object("nalgonda-cloud-assets", "config.json");
        console.log("✅ Success:", response.data);
    } catch (err) {
        console.error("❌ Failed:", err.message);
    }
}

handler();
```

## 4. Object-Oriented Programming (Classes)
Both Python and JavaScript support classes with constructors and inheritance:

| Feature | Python | JavaScript |
| :--- | :--- | :--- |
| **Constructor** | `def __init__(self, ...):` | `constructor(...) { ... }` |
| **Instance Reference** | `self.attribute` | `this.attribute` |
| **Inheritance** | `class Child(Parent):` | `class Child extends Parent {` |
| **Super Call** | `super().__init__(...)` | `super(...)` |


````python
# Python Class
class StorageBucket:
    def __init__(self, name):
        self.name = name
    def upload(self, file):
        return f"Uploaded {file} to {self.name}"

b = StorageBucket("nalgonda-s3")
print(b.upload("index.html"))
````

### The JavaScript Equivalent:
```javascript
class StorageBucket {
    constructor(name) {
        this.name = name;
    }
    
    upload(file) {
        return `Uploaded ${file} to ${this.name}`;
    }
}

const b = new StorageBucket("nalgonda-s3");
console.log(b.upload("index.html"));
```
